```python
# 파이썬 유틸리티 함수 정의
import matplotlib.pyplot as plt
import numpy as np
  
def plot_image(image, title=''):
    plt.imshow(image)
    plt.axis('off') 
    plt.title(title, size=20) 
    
def plot_comp_image(image1, image2, title1='', title2=''):    
    plt.subplot(121), plt.imshow(image1, cmap='gray'), plt.axis('off')  
    plt.title(title1, size=20) 
    plt.subplot(122), plt.imshow(image2, cmap='gray'), plt.axis('off')   
    plt.title(title2, size=20)    
    
def plot_hist(r, g, b, title=''):
    plt.hist(np.array(r).ravel(), bins=256, range=(0, 256), color='r', alpha=0.5)
    plt.hist(np.array(g).ravel(), bins=256, range=(0, 256), color='g', alpha=0.5)
    plt.hist(np.array(b).ravel(), bins=256, range=(0, 256), color='b', alpha=0.5)
    plt.xlabel('pixel value', size=20), plt.ylabel('frequency', size=20)
    plt.title(title, size=20)    
    
def plot_gray_hist(image, title=''):
    plt.hist(np.array(image).ravel(), bins=256, range=(0, 256), color='gray', alpha=1.0)
    plt.xlabel('pixel value', size=20), plt.ylabel('frequency', size=20)
    plt.title(title, size=20)
```

```python
# 히스토그램 보여주기
from PIL import Image 
import matplotlib.pyplot as plt

im = Image.open("./drive/MyDrive/Images/parrot.png")
im_r, im_g, im_b = im.split() # 각 채널로 나누기

plt.figure(figsize=(15,5))
plt.subplot(121), plot_image(im, 'original image')
plt.subplot(122), plot_hist(im_r, im_g, im_b,'histogram for RGB channels')
#plt.subplot(122), plot_gray_hist(im_r,'histogram for Red channels')
plt.show()
```

```python
# 선형대비 늘리기 
from PIL import Image 
import matplotlib.pyplot as plt

def ContrastStretch(px): # 선형대비 늘리기 함수
    r1, r2 = 10, 70
    if px < r1 : return 0
    elif px > r2 : return 255
    else: return 255*(px - r1) / (r2-r1)

im = Image.open("./drive/MyDrive/Images/beans.jpg")
im_c = im.point(ContrastStretch) # 화소별로 함수를 적용
plt.figure(figsize=(10,5))
plot_comp_image(im, im_c, 'Low contrast', 'Contrast stretching')
# 히스토그램 출력하여 비교해 보기
```
