```python
# 파이썬 유틸리티 함수 정의
import matplotlib.pyplot as plt
import numpy as np
  
def plot_image(image, title=''):
    plt.imshow(image)
    plt.axis('off') 
    plt.title(title, size=20) 
    
def plot_gray_hist(image, title=''):
    plt.hist(np.array(image).ravel(), bins=256, range=(0, 256), color='gray', alpha=1.0)
    plt.xlabel('pixel value', size=20), plt.ylabel('frequency', size=20)
    plt.title(title, size=20)
```

```python
# 히스토그램 평활화
from PIL import Image 
import matplotlib.pyplot as plt
from skimage import exposure, io  # skimage 라이브러리에 다양한 함수 제공

im = io.imread("./drive/MyDrive/Images/beans.jpg")
#im = np.array(Image.open("./drive/MyDrive/Images/beans.jpg"))
im_eq = exposure.equalize_hist(im) # 결과가 [0~1]

plt.figure(figsize=(15,10))
plt.subplot(221), plot_image(im, 'Source')
plt.subplot(222), plot_image(im_eq, 'Equalized')
plt.subplot(223), plot_gray_hist(im, 'Source histogram')
plt.subplot(224), plot_gray_hist(im_eq*255, 'Equalized histogram') # 화면에 보여주기 위해 크기 조정
plt.show()
```

```python
# 히스토그램 매칭
import matplotlib.pyplot as plt
from skimage import exposure, io, color

src = io.imread("./drive/MyDrive/Images/beans.jpg")
ref = io.imread("./drive/MyDrive/Images/lena.jpg")
matched = exposure.match_histograms(src, ref)

plt.figure(figsize=(15,10))
plt.subplot(231), plot_image(src, 'Source')
plt.subplot(232), plot_image(ref, 'Reference')
plt.subplot(233), plot_image(color.rgb2gray(matched), 'Matched')
plt.subplot(234), plot_gray_hist(src, 'Source')
plt.subplot(235), plot_gray_hist(ref, 'Reference')
plt.subplot(236), plot_gray_hist(matched, 'Matched')
plt.show()
```
