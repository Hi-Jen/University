```python
# 파이썬 유틸리티 함수 정의
import matplotlib.pyplot as plt
  
def plot_image(image, title=''):
    plt.imshow(image)
    plt.axis('off') 
    plt.title(title, size=20) 
```

```python
# 컬러변환
import matplotlib.pyplot as plt
from skimage import io, color

im = io.imread("./drive/MyDrive/Images/colorblind.png")
im1 = (im[:,:,0] + im[:,:,1] + im[:,:,2]) / 3  # 가장 간단한 채널 평균
im2 = (im[:,:,0] + 2.0*im[:,:,1] + im[:,:,2]) / 4  # 녹색을 강조한 변환
im3 = color.rgb2gray(im)  
# im3 = color.rgb2hsv(im) # 다양한 채널 색상 변환 함수
plt.figure(figsize=(10,10))
plt.gray()
plt.subplot(221), plot_image(im, 'Source')
plt.subplot(222), plot_image(im1, 'Simple')
plt.subplot(223), plot_image(im2, 'Adaptive')
plt.subplot(224), plot_image(im3, 'Library')
plt.show()
```

```python
# 크로마킹
import matplotlib.pyplot as plt
from skimage import io
import numpy as np

fore = io.imread("./drive/MyDrive/Images/superman.jpg")
back = io.imread("./drive/MyDrive/Images/building.jpg")

b, g, r = fore[:, :, 0], fore[:,:,1], fore[:,:,2] # 각 채널로 분류
mask = (g > 100) & (r < 100) & (b < 100) # 녹색과 관련 내용만 걸러내기

foremask = np.copy(fore)
foremask[mask != 0] = [0, 0, 0]

backmask = np.copy(back)
backmask[mask == 0] = [0, 0, 0]

chroma = foremask + backmask # 마스크에 의한 배경과 전경을 합하기

plt.figure(figsize=(15,10))
plt.subplot(221), plot_image(fore, 'Foreground')
plt.subplot(222), plot_image(back, 'Background') 
plt.subplot(223), plot_image(mask, 'Masking')
plt.subplot(224), plot_image(chroma, 'Result')
plt.show()
```
