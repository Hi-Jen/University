```python
# 파이썬 유틸리티 함수 정의
import matplotlib.pyplot as plt
import numpy as np

def plot_image(image, title=''):
    plt.imshow(image)
    plt.axis('off')
    plt.title(title, size=20)
```

```python
# Boundary(경계)
from skimage.morphology import binary_erosion, square
import matplotlib.pyplot as plt
from skimage import io, color

im = color.rgb2gray(color.rgba2rgb(io.imread('dynasaur.png')))
im[im <= 0.5] = 0
im[im > 0.5] = 1

boundary = im - binary_erosion(im, square(5))
boundary[boundary <= 0.5] = 0
boundary[boundary > 0.5] = 1

plt.figure(figsize=(10,5))
plt.gray()
plt.subplot(121), plot_image(im, 'Source')
plt.subplot(122), plot_image(boundary, 'Boundary')
plt.show()
```

```python
# Skeleton(골격화)
from skimage.morphology import skeletonize
import matplotlib.pyplot as plt
from skimage import io, color

im = color.rgb2gray(color.rgba2rgb(io.imread('dynasaur.png')))
im[im <= 0.5] = 0
im[im > 0.5] = 1
im = 1 - im
skeleton = skeletonize(im)

plt.figure(figsize=(10,5))
plt.gray()
plt.subplot(121), plot_image(im, 'Source')
plt.subplot(122), plot_image(skeleton, 'Skeleton')
plt.show()
```

```python
# HMT(Hit-Mist-Transformation)
from skimage.morphology import binary_erosion, binary_dilation, binary_opening, binary_closing, rectangle, disk, square
import matplotlib.pyplot as plt
from skimage import io, color
from skimage.draw import circle_perimeter

im1 = color.rgb2gray(io.imread('HitMissInput.png'))
im2 = color.rgb2gray(io.imread('HitMissSE.png'))

im1[im1 <= 0.5] = 0
im1[im1 > 0.5] = 1
im2[im2 <= 0.5] = 0
im2[im2 > 0.5] = 1

A = im1.copy()
NotA = 1 - A

Bon = im2.copy()
Boff = 1 - Bon

find1 = binary_erosion(A, Bon)
find2 = binary_erosion(NotA, Boff)

idx1_find = np.where(find1 == 1)
idx1_set = list(zip(idx1_find[0], idx1_find[1]))

idx2_find = np.where(find2 == 1)
idx2_set = list(zip(idx2_find[0], idx2_find[1]))

comp_set = set(idx1_set).intersection(idx2_set)
result_set = list(comp_set)

im_result = np.stack((im1, im1, im1), axis=2)

for (row, col) in result_set:
    rr, cc = circle_perimeter(row, col, (int)(im2.shape[0]/2))
    im_result[rr, cc] = (1, 0, 0)
    rr, cc = circle_perimeter(row, col, (int)(im2.shape[0]/2)+1)
    im_result[rr, cc] = (1, 0, 0)

plt.figure(figsize=(20,5))
plt.subplot(121), plot_image(im2, 'SE')
plt.subplot(122), plot_image(im_result, 'Result')
plt.show()
```
