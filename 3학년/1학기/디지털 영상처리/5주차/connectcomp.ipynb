```python
# 파이썬 유틸리티 함수 정의
import matplotlib.pyplot as plt
import numpy as np

def plot_image(image, title=''):
    plt.imshow(image)
    plt.axis('off')
    plt.title(title, size=20)
```

```python
# 홍수 채움(fill holes)
from skimage.morphology import reconstruction
import matplotlib.pyplot as plt
from skimage import io, color

im = color.rgb2gray(color.rgba2rgb(io.imread('text.png')))
seed = np.copy(im)
seed[1:-1, 1:-1] = im.max()
mask = im

filled = reconstruction(seed, mask, method='erosion')
#filled[filled <= 0.2] = 0
#filled[filled > 0.2] = 1

plt.figure(figsize=(10,5))
plt.gray()
plt.subplot(121), plot_image(im, 'Source')
plt.subplot(122), plot_image(filled, 'Fill Holes')
plt.show()
```

```python
# Connected Component Labeling(연결 요소 레이블링)
from skimage.filters import threshold_otsu
from skimage.morphology import binary_closing, square
from skimage.measure import label
import matplotlib.pyplot as plt
from skimage import io, color

im = io.imread('coins.jpg')

thresh = threshold_otsu(im)
im_bw =  np.copy(im)
im_bw[im <= thresh] = 0
im_bw[im > thresh] = 1
im_bw = binary_closing(im_bw, square(5))

label_image = label(im_bw)
label_overlay = color.label2rgb(label_image, image=im, bg_label=0)

plt.figure(figsize=(10,10))
plt.gray()
plt.subplot(221), plot_image(im, 'Source')
plt.subplot(222), plot_image(im_bw, 'Threshold')
plt.subplot(223), plot_image(label_image, 'Labeled')
plt.subplot(224), plot_image(label_overlay, 'Colored')
plt.show()
```
