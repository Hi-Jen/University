```python
# 파이썬 유틸리티 함수 정의
import matplotlib.pyplot as plt
import numpy as np

def plot_image(image, title=''):
    plt.imshow(image)
    plt.axis('off')
    plt.title(title, size=20)
```

```python
# 지문 영상 모폴로지 연산
from skimage.morphology import binary_opening, binary_closing, square
import matplotlib.pyplot as plt
from skimage import io, color

im = color.rgb2gray(io.imread('fingerprint.jpg'))

im[im <= 0.5] = 0
im[im > 0.5] = 1

im_o = binary_opening(im, square(3))
im_c = binary_closing(im, square(3))
im_oc = binary_closing(binary_opening(im, square(2)), square(3))

plt.figure(figsize=(10,10))
plt.gray()
plt.subplot(221), plot_image(im, 'original')
plt.subplot(222), plot_image(im_o, 'opening')
plt.subplot(223), plot_image(im_c, 'closing')
plt.subplot(224), plot_image(im_oc, 'opening + closing')
plt.show()
```

```python
# Convex Hull (볼록체)
from skimage.morphology import convex_hull_image
import matplotlib.pyplot as plt
from skimage import io, color

im = color.rgb2gray(io.imread('horse-dog.jpg'))
im[im < 0.5] = 0
im[im >= 0.5] = 1
chull = convex_hull_image(im)

plt.figure(figsize=(10,10))
plt.gray()
plt.subplot(121), plot_image(im, 'Source')
plt.subplot(122), plot_image(chull, 'Convex Hull')
plt.show()
```
