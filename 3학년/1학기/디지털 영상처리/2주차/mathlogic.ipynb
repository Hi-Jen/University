```python
# 화소 변환
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np  # 화소를 접근하기 위한 배열 데이터 형식

im_beans = Image.open("/content/drive/MyDrive/Colab Notebooks/week02/beans.jpg")
im_gainbias = im_beans.point(lambda x: float(x)*2.5 +50)  # 바이어스와 게인

im_skull = Image.open("/content/drive/MyDrive/Colab Notebooks/week02/skull.jpg")
im_inv = im_skull.point(lambda x: 255 - x) # 반전

im_dft = Image.open("/content/drive/MyDrive/Colab Notebooks/week02//dft.jpg")
im_log = im_dft.point(lambda x: 50*np.log(1+float(x))) # 분석변환 중 로그함수

im_spine = Image.open("/content/drive/MyDrive/Colab Notebooks/week02/spine.jpg")
im_pow = im_spine.point(lambda x: 255*(x/255)**0.3) # 분석 변환 중 지수함수

plt.figure(figsize=(10,10)) # 화면 크기(단위 인치), 해상도 기본 100 dpi
plt.gray() # 전체 회색조로 출력
plt.subplot(141), plt.imshow(im_beans), plt.axis('off')  #화면을 나누어 출력
plt.subplot(142), plt.imshow(im_skull), plt.axis('off')
plt.subplot(143), plt.imshow(im_dft), plt.axis('off')
plt.subplot(144), plt.imshow(im_spine), plt.axis('off')

plt.subplot(245), plt.imshow(im_gainbias), plt.axis('off')
plt.subplot(246), plt.imshow(im_inv), plt.axis('off')
plt.subplot(247), plt.imshow(im_log), plt.axis('off')
plt.subplot(248), plt.imshow(im_pow), plt.axis('off')
plt.show()
```

```python
# 영상간의 연산
from PIL import Image
from PIL.ImageChops import add, subtract, multiply, difference # PIL 내에 다양한 연산
import matplotlib.pylab as plt

im1 = Image.open("/content/drive/MyDrive/Colab Notebooks/week02/goal1.png")
im2 = Image.open("/content/drive/MyDrive/Colab Notebooks/week02/goal2.png")
im = difference(im1, im2) # 배경 절대값 감산 연산

plt.figure(figsize=(10, 10))
plt.subplot(311), plt.imshow(im1), plt.axis('off')
plt.subplot(312), plt.imshow(im2), plt.axis('off')
plt.subplot(313), plt.imshow(im), plt.axis('off')
plt.show()
```
