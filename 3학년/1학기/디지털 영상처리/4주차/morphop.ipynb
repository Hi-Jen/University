```python
# 파이썬 유틸리티 함수 정의
import matplotlib.pyplot as plt
import numpy as np

def plot_image(image, title=''):
    plt.imshow(image)
    plt.axis('off')
    plt.title(title, size=20)
```

```python
# Erosion(침식)
from skimage.morphology import binary_erosion, binary_dilation, binary_opening, binary_closing, rectangle, disk, square
import matplotlib.pyplot as plt
from skimage import io, color

im = color.rgb2gray(io.imread('clock2.jpg'))
im[im <= 0.5] = 0  # 이진 영상 만들기
im[im > 0.5] = 1

im1 = binary_erosion(im, rectangle(1,5))
im2 = binary_erosion(im, rectangle(1,15))

plt.figure(figsize=(20,10))
plt.gray()
plt.subplot(131), plot_image(im, 'original')
plt.subplot(132), plot_image(im1, 'erosion: size(1,5)')
plt.subplot(133), plot_image(im2, 'erosion: size(1,15)')
plt.show()
```

```python
# Dilation(팽창)
from skimage.morphology import binary_erosion, binary_dilation, binary_opening, binary_closing, rectangle, disk, square
import matplotlib.pyplot as plt
from skimage import io, color

im = color.rgb2gray(color.rgba2rgb(io.imread('tagore.png')))
im[im <= 0.5] = 0
im[im > 0.5] = 1

im1 = binary_dilation(im, disk(2))
im2 = binary_dilation(im, disk(4))

plt.figure(figsize=(20,10))
plt.gray()
plt.subplot(131), plot_image(im, 'original')
plt.subplot(132), plot_image(im1, 'dilation: disk(2)')
plt.subplot(133), plot_image(im2, 'dilation: disk(4)')
plt.show()
```

```python
# Opening: 열림(침식+팽창) 닫힘(팽창+침식)

from skimage.morphology import binary_erosion, binary_dilation, binary_opening, binary_closing, rectangle, disk, square
import matplotlib.pyplot as plt
from skimage import io, color

im = color.rgb2gray(io.imread('circles.jpg'))
im[im <= 0.5] = 0
im[im > 0.5] = 1

im1 = binary_opening(im, disk(12))
im2 = binary_closing(im, disk(6))

plt.figure(figsize=(20,10))
plt.gray()
plt.subplot(1,3,1), plot_image(im, 'original')
plt.subplot(1,3,2), plot_image(im1, 'opening disk size 12')
plt.subplot(1,3,3), plot_image(im2, 'closing disk size 6')
plt.show()
```
