```python
# 파이썬 유틸리티 함수 정의
import matplotlib.pyplot as plt
import numpy as np

def plot_image(image, title=''):
    plt.imshow(image)
    plt.axis('off')
    plt.title(title, size=20)
```

```python
# 구조요소 만들기
from skimage.morphology import rectangle, disk, square
import matplotlib.pyplot as plt

se1 = rectangle(7, 11)  # 직사각형
se2 = square(11) # 정사각형
#se2[5-3+1:5+3, 5-3+1:5+3] = 255
se3 = disk(5) # 원

plt.figure(figsize=(15,5))
plt.gray()
plt.subplot(131), plot_image(se1, 'Rectangle')
plt.subplot(132), plot_image(se2, 'Square')
plt.subplot(133), plot_image(se3, 'disk')
plt.show()
```

```python
# 배열을 활용한 슬라이싱
import matplotlib.pyplot as plt
from skimage import io, filters
import numpy as np

img = io.imread("parrot.png")

lx, ly, _ = img.shape
X, Y = np.ogrid[0:lx, 0:ly] # 마스킹을 위한 범위 설정 인덱스
mask = (X - lx / 2) ** 2 + (Y - ly / 2) ** 2 > lx * ly / 4

img_mask = np.copy(img)
img_mask[mask,:] = 0  # 원 모양
#img_mask[10:100, 100:200, :] = 0 # 사각형 모양

plt.figure(figsize=(10,5))
plt.subplot(121), plot_image(img, 'Source')
plt.subplot(122), plot_image(img_mask, 'Slicing')
plt.show()
```
