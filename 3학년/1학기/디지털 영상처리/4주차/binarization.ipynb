```python
# 파이썬 유틸리티 함수 정의
import matplotlib.pyplot as plt
import numpy as np

def plot_image(image, title=''):
    plt.imshow(image)
    plt.axis('off')
    plt.title(title, size=20)

def plot_gray_hist(image, title=''):
    plt.hist(np.array(image).ravel(), bins=256, range=(0, 256), color='gray', alpha=1.0)
    plt.xlabel('pixel value', size=20), plt.ylabel('frequency', size=20)
    plt.title(title, size=20)
```

```python
# 이진영상 생성
from PIL import Image
import matplotlib.pyplot as plt
from skimage import io, color, filters
import numpy as np

im = color.rgb2gray(io.imread('cells.jpg')) # [0.0 ~ 1.0]

# 임계치 설정
im_bin = np.copy(im)
threshold = 0.5  # 임의로 정의된 값으로 임계치 설정
#threshold = filters.threshold_otsu(im) # 오츄 알고리즘에 의한 최적치 찾기
im_bin[im <= threshold] = 0
im_bin[im > threshold] = 1

plt.figure(figsize=(20,5))
plt.gray()
plt.subplot(131), plot_image(im, 'Gray')
plt.subplot(132), plot_gray_hist(im*255, 'Historgram')
plt.axvline(threshold*256, color='r')
plt.subplot(133), plot_image(im_bin, 'Binary')
plt.show()
```
